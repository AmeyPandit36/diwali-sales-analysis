# **EDA**

Gender

```python
df.value_counts('Gender')
```

```python
df.groupby('Gender')['Amount'].sum().sort_values(ascending=False)
```

From Above Result We cansee that most of the buyers are Females and the  even the purchasing power of females is greater than men

Age

```python
df.value_counts('Age Group')
```

```python
df.groupby(['Age Group','Gender'])['Amount'].sum().sort_values(ascending=False)
```

From above table we can see that most of the buyers are in age-group 26-35 and are female

**State**

```python
df.columns
```

```python
df.value_counts('State').head(30)
```

```python
df.groupby(['State','Gender'])['Amount'].sum().sort_values(ascending=False)
```
